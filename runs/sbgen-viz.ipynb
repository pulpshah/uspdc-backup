```python
# Start program timer
import time
start_time = time.time()
```

```python
# Import required modules
from transcript_configurator import TranscriptConfigurator
from requester import GPTConcurrentRequester
from scoreboard import ScoreboardCalculator
from debate_parser import DebateParser
import pandas as pd
import dtale
import spacy
import nltk
import os
```

```python
# Initialize critical objects
open_ai_keys = [os.getenv(f"OPENAI_KEY_{i}") for i in range(1, 26)]
pd.set_option('display.max_columns', 500)
nlp = spacy.load("en_core_web_lg")
english_vocab = set(w.lower() for w in nltk.corpus.words.words())
```

```python
# Load transcripts and speaker names
original_transcript_path = "../runs/2008-09-26-debate-preclean.txt"
known_speakers = {'LEHRER', 'OBAMA', 'MCCAIN'}
```

```python
# Configure transcript scoreboard
configurator = TranscriptConfigurator()
empty_scoreboard = configurator.get_updated_scoreboard(original_transcript_path, known_speakers)
```

```python
# Initalize base scores
scoreboard_calculator = ScoreboardCalculator(empty_scoreboard, nlp, english_vocab)
```

```python
# Request EPL information from GPT
gpt_requester = GPTConcurrentRequester(empty_scoreboard, open_ai_keys, original_transcript_path, scoreboard_calculator)
final_scoreboard = gpt_requester.get_epl()
```

```python
# Update scoreboard values with EPL info
updated_final_scoreboard = scoreboard_calculator.update_scoreboard()
```

```python
# Display using dtale
x = dtale.show(updated_final_scoreboard)
x.open_browser()
```

```python
# print program execution time
elapsed = (time.time() - start_time)
print('Execution time:', time.strftime("%H:%M:%S", time.gmtime(elapsed)))
```
