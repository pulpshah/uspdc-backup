```python
import os
import re
import json

def process_directory(directory):
    for filename in os.listdir(directory):
        if filename.endswith(".txt"):
            with open(os.path.join(directory, filename), 'r') as file:
                data = file.read()

            result = {
                'number_of_lines': len(data.split('\n')),
                'number_of_blank_lines': len([line for line in data.split('\n') if line.strip() == '']),
                'number_of_sentences': len(re.findall(r'[.!?]', data)),
                'number_of_words': len(data.split()),
                'number_of_chars_excluding_spaces': len(data.replace(' ', '')),
                'number_of_chars_including_spaces': len(data),
                'number_of_ellipses': len(re.findall(r'(\.\s*){2,}', data)),
                'number_of_dashes': len(re.findall(r' - | --', data)),
                'number_of_exclamation_marks': data.count('!'),
                'number_of_question_marks': data.count('?'),
                'number_of_words_in_brackets': len(re.findall(r'\[.*?\]|\(.*?\)', data)),
                'number_of_other_special_chars': len(re.findall(r'[^a-zA-Z0-9\s\.\-!?]', data)),
                'number_of_all_caps_words': len([word for word in data.split() if word.isupper()]),
            }

            with open(os.path.join(directory, filename.replace('.txt', '.json')), 'w') as json_file:
                json.dump(result, json_file, indent=4)

```
